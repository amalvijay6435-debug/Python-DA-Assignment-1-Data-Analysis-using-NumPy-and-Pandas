**Creating Pandas Series**

 Import pandas and create a series name as mark and ranks print the result

Use the integer index position to access the mark of the 1st rank student.

```python
mark =pd.Series([95,92,89,85,80],index=["Rank1","Rank2","Rank3","Rank4","Rank5"])
mark =mark[1]
print(mark)
```

Using the loc function to find the first three rank and print the result.

```python
mark =pd.Series([95,92,89,85,80],index=["Rank1","Rank2","Rank3","Rank4","Rank5"])
print(mark.loc[["Rank1","Rank2","Rank3"]])
```

```python
mark =pd.Series([95,92,89,85,80],index=["Rank1","Rank2","Rank3","Rank4","Rank5"])
print(mark.iloc[2])
```

Manipulating the DataFrame

Modify the 'Amount' for the transaction with TransactionID 102 to 165

create a small dataframe using df,loc function which transaction =102 and amount =102 and print the output

```python
df = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104],
    'Region': ['North', 'South', 'North', 'East'],
    'Amount': [250, 300, 150, 800]
})

df.loc[df['TransactionID'] == 102, 'Amount'] = 165

print(df)
```

Add a new column 'Discount' by calculating 10% of the 'Amount'.

Add new column discount in the dataframe and amount price divide by 10 to discount column to print the output

```python
df = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104],
    'Region': ['North', 'South', 'North', 'East'],
    'Amount': [250, 300, 150, 800]
})
df['Discount']=df['Amount']/10
print(df)
```

Remove the row with TransactionID 109.

Remove the transactionid 109 in the column to dataframe not equal to 109 and print the result.


```python
df = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104],
    'Region': ['North', 'South', 'North', 'East'],
    'Amount': [250, 300, 150, 800]
})
df = df[df['TransactionID']!=109]
print(df)
```

Delete the 'Discount' column from the DataFrame.


Using drop function to delete the discount column from dataframe and print the output.

```python
df = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104],
    'Region': ['North', 'South', 'North', 'East'],
    'Amount': [250, 300, 150, 800],
    'Discount':[25.0,30.0,15.0,80.0]
})
df= df.drop(columns='Discount')
print(df)
```
